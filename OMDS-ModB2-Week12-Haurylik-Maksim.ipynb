# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
# Figure 8.1

products = ["Product A", "Product B", "Product C", "Product D", "Product E"]

# My approximation:
data = {
    "2019": [395, 360,   0,   0,   0],
    "2020": [420, 400,   0,   0,   0],
    "2021": [425, 408, 100,   0,   0],
    "2022": [390, 375, 180, 160,   0],
    "2023": [300, 290, 200, 260,   0],
    "2024": [270, 260, 240, 220, 100],
    "2025": [260, 250, 180, 215, 210],
}

years = list(data.keys())
values = np.array(list(data.values()))

x = np.arange(len(products))
width = 0.11

fig, ax = plt.subplots(figsize=(9, 5.8))

# More or less those colors
colors = [
    "#86ABDC",  # 2019
    "#E48888",  # 2020
    "#B4C586",  # 2021
    "#B7A7CF",  # 2022
    "#7EBDD8",  # 2023
    "#F8AE71",  # 2024
    "#AABAD4",  # 2025
]

for i, year in enumerate(years):
    ax.bar(x + (i - 3) * width, values[i], width=width, label=year, color=colors[i], edgecolor=colors[i])

ax.set_title("Average Retail Product Price Per Year", fontsize=19, weight="bold", pad=18)
ax.set_xticks(x)
ax.set_xticklabels(products, fontsize=12)
ax.set_ylim(0, 500)
ax.set_yticks(np.arange(0, 501, 100))
ax.set_yticklabels([f"${v}" for v in range(0, 501, 100)], fontsize=11)
ax.yaxis.grid(True, color="#d0d0d0", linewidth=1)
ax.set_axisbelow(True)

ax.legend(
    ncol=7,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.12),
    frameon=False,
    fontsize=10,
    handlelength=0.6,
    handletextpad=0.4
)

plt.tight_layout(rect=[0.02, 0.06, 0.98, 0.95])
plt.show()
```
